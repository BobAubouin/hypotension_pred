```python
from pathlib import Path
from operator import xor
import pickle

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from scipy.stats import shapiro

import hp_pred.experiments as expe



def print_metric(array):
    print(f"{np.mean(array):.3f} (95% CI, {np.quantile(array, 0.025):.3f} to {np.quantile(array, 0.975):.3f})")

with open('data/results/baseline_chu_test_no_hypo.pkl', 'rb') as f:
    baseline_full = pickle.load(f)

with open('data/results/baseline_chu_test_no_hypo_no_inter.pkl', 'rb') as f:
    baseline_no_hypo_no_inter = pickle.load(f)


auc_op_full = baseline_full["au_op_prcs"]
auc_op_no_hypo_no_inter = baseline_no_hypo_no_inter["au_op_prcs"]

# t-test for AUC OP
print_metric(auc_op_full - auc_op_no_hypo_no_inter)
```

```python
with open('data/results/xgb_inter_easy_points_filtered_chu_test_no_hypo_no_inter.pkl', 'rb') as f:
    model_NF_IF = pickle.load(f)

with open('data/results/xgb_easy_point_filtered_chu_test_no_hypo_no_inter.pkl', 'rb') as f:
    model_NF = pickle.load(f)


auc_op_NF_IF = model_NF_IF["auprcs"]
auc_op_NF = model_NF["auprcs"]

print_metric(auc_op_NF_IF) 
print_metric(auc_op_NF) 
print_metric(auc_op_NF_IF - auc_op_NF) 
```

```python
with open('data/results/xgb_inter_easy_points_filtered_chu_test_no_hypo_no_inter.pkl', 'rb') as f:
    model_NF_IF = pickle.load(f)

with open('data/results/xgb_inter_filtered_chu_test_no_hypo_no_inter.pkl', 'rb') as f:
    model_IF = pickle.load(f)


auc_op_NF_IF = model_NF_IF["au_op_prcs"]  # au_op_prcs, auprcs, aucs
auc_op_IF = model_IF["au_op_prcs"]

print_metric(auc_op_NF_IF) 
print_metric(auc_op_IF) 
print_metric(auc_op_IF - auc_op_NF_IF) 
```

```python
metric = "au_op_prcs" # au_op_prcs, auprcs, aucs
with open('data/results/xgb_no_filter_chu_test_no_hypo.pkl', 'rb') as f:
    model_All_no_hypo_no_inter = pickle.load(f)

with open('data/results/xgb_easy_point_filtered_chu_test_no_hypo.pkl', 'rb') as f:
    model_NF_no_hypo_no_inter = pickle.load(f)
with open('data/results/xgb_inter_easy_points_filtered_chu_test_no_hypo.pkl', 'rb') as f:
    model_NF_IF_no_hypo_no_inter = pickle.load(f)

with open('data/results/xgb_inter_filtered_chu_test_no_hypo.pkl', 'rb') as f:
    model_IF_no_hypo_no_inter = pickle.load(f)

auc_op_All_no_hypo_no_inter = model_All_no_hypo_no_inter[metric]
auc_op_NF_no_hypo_no_inter = model_NF_no_hypo_no_inter[metric]
auc_op_NF_IF_no_hypo_no_inter = model_NF_IF_no_hypo_no_inter[metric]
auc_op_IF_no_hypo_no_inter = model_IF_no_hypo_no_inter[metric]

no_hyo = np.concatenate((auc_op_All_no_hypo_no_inter,auc_op_NF_no_hypo_no_inter,auc_op_NF_IF_no_hypo_no_inter,auc_op_IF_no_hypo_no_inter))

print_metric(no_hyo)
print('Mean performance on no_hypo')
p_1= np.mean(no_hyo)



with open('data/results/xgb_no_filter_chu_test_full.pkl', 'rb') as f:
    model_All_no_hypo_no_inter = pickle.load(f)

with open('data/results/xgb_easy_point_filtered_chu_test_full.pkl', 'rb') as f:
    model_NF_no_hypo_no_inter = pickle.load(f)
with open('data/results/xgb_inter_easy_points_filtered_chu_test_full.pkl', 'rb') as f:
    model_NF_IF_no_hypo_no_inter = pickle.load(f)

with open('data/results/xgb_inter_filtered_chu_test_full.pkl', 'rb') as f:
    model_IF_no_hypo_no_inter = pickle.load(f)

auc_op_All_no_hypo_no_inter = model_All_no_hypo_no_inter[metric]
auc_op_NF_no_hypo_no_inter = model_NF_no_hypo_no_inter[metric]
auc_op_NF_IF_no_hypo_no_inter = model_NF_IF_no_hypo_no_inter[metric]
auc_op_IF_no_hypo_no_inter = model_IF_no_hypo_no_inter[metric]

print('Mean performance on full')
no_hyo = np.concatenate((auc_op_All_no_hypo_no_inter,auc_op_NF_no_hypo_no_inter,auc_op_NF_IF_no_hypo_no_inter,auc_op_IF_no_hypo_no_inter))
_,p = shapiro(no_hyo)
p2= np.mean(no_hyo)
print_metric(no_hyo)

print(f"{np.abs(p_1-p2)/p_1:.1%}")

```
